```python
# Mengimpor library NumPy untuk memudahkan operasi matriks dan perhitungan numerik
import numpy as np

# Mendefinisikan matriks augmented (ukuran 3x4) dari contoh SPLTV:
# Persamaan 1: 2x +  y -  z =  8
# Persamaan 2: -3x - y + 2z = -11
# Persamaan 3: -2x + y + 2z = -3
# Tipe data float64 digunakan agar hasil pembagian desimal tetap presisi
A_aug = np.array([
    [ 2.0,  1.0, -1.0,   8.0],
    [-3.0, -1.0,  2.0, -11.0],
    [-2.0,  1.0,  2.0,  -3.0]
], dtype=np.float64)

# Menampilkan matriks awal ke layar sebelum dilakukan proses eliminasi
print("Matriks Augmented Awal:")
print(A_aug)
```

```python
# Mendefinisikan fungsi eliminasi maju untuk mengubah matriks menjadi bentuk segitiga atas
def eliminasi_maju(matriks):
    # Membuat salinan matriks agar data asli di Cell 1 tidak ikut berubah
    M = matriks.copy()
    # Mengambil jumlah baris matriks (dalam kasus SPLTV, n = 3)
    n = len(M)

    # Melakukan iterasi untuk setiap kolom pivot (kolom 0 dan kolom 1)
    for i in range(n - 1):
        # --- TAHAP PARTIAL PIVOTING ---
        # Mencari indeks baris dengan nilai absolut terbesar pada kolom ke-i mulai dari baris i ke bawah
        baris_max = i + np.argmax(np.abs(M[i:, i]))

        # Jika baris dengan nilai terbesar bukan baris pivot saat ini, tukar posisi kedua baris tersebut
        if baris_max != i:
            M[[i, baris_max]] = M[[baris_max, i]]

        # Memeriksa apakah elemen pivot bernilai nol (menandakan matriks singular / tidak punya solusi unik)
        if np.isclose(M[i, i], 0.0):
            raise ValueError("Pivot bernilai nol: Sistem persamaan tidak memiliki solusi unik.")

        # --- TAHAP ELIMINASI BARIS ---
        # Melakukan perulangan untuk setiap baris j yang berada di bawah baris pivot i
        for j in range(i + 1, n):
            # Menghitung faktor pengali agar elemen di bawah pivot menjadi nol
            faktor = M[j, i] / M[i, i]
            # Mengurangi baris ke-j dengan hasil perkalian faktor dan baris pivot ke-i
            M[j, i:] = M[j, i:] - faktor * M[i, i:]

    # Mengembalikan matriks yang sudah berbentuk segitiga atas (Upper Triangular Matrix)
    return M

# Menjalankan fungsi eliminasi maju pada matriks awal dan menyimpan hasilnya
matriks_segitiga = eliminasi_maju(A_aug)

# Menampilkan matriks hasil eliminasi maju
print("Matriks Setelah Eliminasi Maju (Segitiga Atas):")
print(np.round(matriks_segitiga, 4))
```

```python
# Mendefinisikan fungsi substitusi mundur untuk mencari nilai variabel x, y, dan z
def substitusi_mundur(M):
    # Mengambil jumlah persamaan (n = 3)
    n = len(M)
    # Memeriksa elemen pivot terakhir; jika nol, sistem tidak memiliki solusi tunggal
    if np.isclose(M[n - 1, n - 1], 0.0):
        raise ValueError("Sistem persamaan inkonsisten atau memiliki tak hingga solusi.")

    # Menyiapkan array kosong berukuran 3 untuk menampung hasil [x, y, z]
    solusi = np.zeros(n, dtype=np.float64)

    # Melakukan perulangan mundur dari baris terbawah (indeks 2) hingga baris teratas (indeks 0)
    for i in range(n - 1, -1, -1):
        # Menghitung jumlah perkalian koefisien dan nilai variabel yang sudah diketahui di sebelah kanan
        jumlah_kanan = np.dot(M[i, i+1:n], solusi[i+1:n])
        # Menghitung nilai variabel ke-i dengan membagi sisa konstanta terhadap elemen diagonal utama
        solusi[i] = (M[i, n] - jumlah_kanan) / M[i, i]

    # Mengembalikan array berisi nilai [x, y, z]
    return solusi

# Menjalankan fungsi substitusi mundur menggunakan matriks segitiga atas dari Cell 2
hasil_akhir = substitusi_mundur(matriks_segitiga)

# Menampilkan nilai masing-masing variabel secara rapi
print("Hasil Penyelesaian SPLTV:")
print(f"x = {hasil_akhir[0]:.4f}")
print(f"y = {hasil_akhir[1]:.4f}")
print(f"z = {hasil_akhir[2]:.4f}")
```

```python
# Memisahkan matriks koefisien (3 kolom pertama) dan vektor konstanta (kolom terakhir)
A = A_aug[:, :3]
B = A_aug[:, 3]

# Menghitung solusi pembanding menggunakan solver linear algebra bawaan NumPy
verifikasi = np.linalg.solve(A, B)

# Menampilkan hasil verifikasi untuk memastikan algoritma manual sudah 100% akurat
print("Verifikasi dengan np.linalg.solve:")
print(f"x = {verifikasi[0]:.4f}, y = {verifikasi[1]:.4f}, z = {verifikasi[2]:.4f}")
```
